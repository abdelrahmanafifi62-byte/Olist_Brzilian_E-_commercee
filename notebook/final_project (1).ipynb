Products cleaning


```python
products=pd.read_csv('olist_products_dataset.csv')
```

```python
products.head(10)
```

```python
products.info()
```

```python
products.isnull().sum()
```

```python
products.rename(
    columns={
        "product_name_lenght": "product_name_length",
        "product_description_lenght": "product_description_length",
    },
    inplace=True,
)
```

```python
products["product_category_name"] = products["product_category_name"].fillna("unknown")
products["product_photos_qty"] = products["product_photos_qty"].fillna(0)
products["product_name_length"] = products["product_name_length"].fillna(0)
products["product_description_length"] = products["product_description_length"].fillna(0)
```

```python
products.isnull().sum()
```

```python
import pandas as pd
all_cols = [
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm',
]


for col in all_cols:
    products[col] = products[col].fillna(
        products.groupby('product_category_name')[col].transform('median')
    )


    products[col] = products[col].fillna(products[col].median())
```

```python
products.isnull().sum()
```

```python
products.duplicated().sum()
```

```python
products['product_id'].duplicated().sum()
```

```python
products.to_csv('cleaned_products.csv', index=False, encoding='utf-8-sig')
```
